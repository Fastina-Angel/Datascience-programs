```python
#importing seaborn
import seaborn as sns
```

```python
import matplotlib.pyplot as plt

data={
    "Product": ["Laptop","Phone","Tablet","Laptop","Phone","Tablet"],
    "Sales": [50,80,40,70,90,60],
    "Price":[50000,30000,20000,55000,32000,22000],
    "Month":["Jan","Jan","Feb","Feb","Mar","Mar"]

    }
```

```python
# 1. BARPLOT
# compares  sales of different products

sns.barplot(x="Product",y="Sales",data=data)
plt.show()
```

```python
# 2. COUNTPLOT
# Count how many times each product appear
sns.countplot(x="Product",data=data, color= "pink")
plt.show()
```

```python
# 3. BOXPLOT
# Show thespread of sales and possible outliers
sns.boxplot(y=data["Sales"], color= "cyan")
plt.show()
```

```python
# 4. SCATTER PLOT
# Show the relationship between price ans sales
sns.scatterplot(x="Price",y="Sales",data=data)
plt.show()
```

```python
# 5.LINE PLOT
# Show the sales trend by month
sns.lineplot(x="Month",y="Sales",data=data)
plt.show()
```

```python
# 6. HIST PLOT

sns.histplot(x="Product",y="Sales",data=data)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data=np.random.randn(1000)
plt.figure(figsize=(8, 6))
sns.histplot(data,bins=30, kde=True, color='skyblue',edgecolor='black')

plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Distribution of Random Data using Seaborn')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
#PLOT

x=[2,3,4,5]
y=[30,60,80,40]
plt.plot(x,y)
plt.fill_between(x,y, color='lightgreen')
plt.show()
```

```python
marks=[45,50,52,55,60,62,65,90]
sns.boxplot(x=marks)
plt.show()
```

```python
sales=[
    [10,20,30,35],
    [15,25,35,30],
    [50,30,40,45]
]
sns.heatmap(sales,annot=True,cmap="Purples")
```
