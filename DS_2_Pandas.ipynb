loc and iloc

```python
data=[10,20,30,40,50]
a=pd.Series(data,index=['a','b','c','d','e'])
print(a.loc['d'])  # accessing elements using label
print(a.iloc[2])  # access elements using position
```

```python
#slicing
data1=[11,23,56,44]
a=pd.Series(data1,index=["a","b","c","d"])
print(a.loc["a":"c"])
print(a.iloc[1:4])

b=a[a>30]
print(b)
print(a[a<30])
```

```python
# removing duplicates
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())

print(ar.nunique())  # count of unique values

print(ar.value_counts()) # count of repeated values g
```

```python
data2=[10,None,20,40,50,None,60]
arr2=pd.Series(data2)
print(arr2.isna())  # None values = True

print(arr2.notna()) # None values = False

print(arr2.dropna()) # Excluding null values
```
