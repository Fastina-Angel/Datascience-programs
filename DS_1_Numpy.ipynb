```python
import numpy as np
marks=np.array([78,92,65,88,45,90,72,55])
print(marks)

total=np.sum(marks)
print(total)

avg=np.average(marks)
print(avg)

max=np.max(marks)
print(max)
min=np.min(marks)
print(min)

#find the students who scored above 75
above75=marks[marks>75]
print(above75)

#marks less than or = to 50
below50=marks[marks<=50]
print(below50)

#sorted array
sorted=np.sort(marks)
print(sorted)
```

```python
temp=np.array([32,35,40,28,42,33])

mean=np.average(temp)
print(mean)

max=np.max(temp)
min=np.min(temp)
print(max,min)

replce=temp[temp>=38]=38
print(temp)

temp2=temp.reshape(2,3)
print(temp2)

print(temp2[0]) #first row
print(temp2[0,1])
print(temp2[:,1]) #second column
```

```python
arr=np.array([10,20,30,40,50,60,70,80])

print(arr[0::7]) #first element& last element
print(arr[2:7])  #elements from inx 2 to 6

print(arr[0:7:2])
```

```python
arr1=np.array([10,21,36,40,59,60,73,80])
print(arr1)

print(np.where(arr1%2==0))  #index of even no. (using where, for index)
print(arr1[arr1%2==0])  #even numbers (without where, for elements)

print(np.where(arr1%2!=0)) #index of odd no.  ,,
print(arr1[arr1%2!=0]) #odd numbers
```
